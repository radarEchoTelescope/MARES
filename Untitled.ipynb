```python
import numpy as np
import matplotlib.pyplot as plt
X_0 = 36.08
E_c = 0.0786
rho_ice = 0.92
```

```python
# /* Shower age */
def ShowerAge(X, E): #            // [g/cm^2, GeV]

    return (3*X/X_0) / ((X/X_0)+2*np.log(E/E_c)); # // Unitless
```

```python
# 	/* Ne, number of ionized electrons per unit length */
# 	// N(X,fEnergy)					Number of particles in the cascade
# 	// 2E6           		[eV/ g/cm^2]  Energy loss per ionizing particle (@ 1 GeV).
# 	// 20            		[eV] Electron ionization energy (per electron).
# 	// 1E5 = 2E6/20 		[# ionized electrons/ # HE ionizing particle/ g/cm^2]
# 	// N*1E5  					[# electrons/ g/cm^2]
# 	// N*1E5*rho_ice	 	[#e / cm]
        

def Ne(X, E): #{                // [g/cm^2, GeV]
    N= 0
    Ne = 0
    if (X > 0):
        N = 0.31*np.exp((X/X_0)*(1-1.5*np.log(ShowerAge(X,E))))/np.sqrt(np.log(E/E_c));
        Ne = N * 1E5 * rho_ice;
    return Ne;       #// [#e / cm]
```

```python
Depths = np.arange(0,2E3,10)
n_large = []
n_small = []
for d in Depths:
    n_large.append(Ne(d,1E10))
    n_small.append(Ne(d,10))
n_large = np.array(n_large)
n_small = np.array(n_small)

plt.plot(Depths, n_large)
plt.plot(Depths, n_small*1E9)
plt.plot(Depths, n_small)
# plt.semilogy()
```
